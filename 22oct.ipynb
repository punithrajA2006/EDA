```python
import numpy as np
import pandas as pd
import seaborn as sns
titanic = sns.load_dataset('titanic')
print(titanic)
```

```python
survival_rate = pd.pivot_table(titanic,values='survived',index=['class', 'sex'],columns='alone')
survival_rate
```

```python
pd.pivot_table(titanic,values=['age', 'fare'],index='sex',columns='class',aggfunc='min')
```

```python
data = {'department': ['CSE', 'CSE', 'CSE', 'CSE', 'AIML', 'AIML', 'AIML', 'AIML', 'ECE', 'ECE', 'ECE', 'ECE'],'gender': ['M', 'F', 'M', 'F', 'M', 'F', 'M', 'F', 'M', 'F', 'M', 'F'],'salary': [75000, 72000, 68000, 80000, 85000, 78000, 90000, 82000, 60000, 65000, 58000, 62000]
}
employees = pd.DataFrame(data)
print(employees)
print()

pd.pivot_table(employees,values='salary',index='gender',columns='department',aggfunc='mean')
```

```python
pd.pivot_table(employees,values='salary',index='gender',columns='department',aggfunc='count')
```

```python
pd.pivot_table(employees,values='salary',index='gender',columns='department',margins=True)
```

```python

employees['age'] = [25, 28, 35, 42, 23, 55, 38, 31, 45, 27, 50, 33]


e1 = pd.cut(employees['age'],bins=[20, 30, 40, 60],labels=['Junior', 'Mid', 'Senior'])
pd.pivot_table(employees,values='salary',index=e1,columns='department')
```

```python
pd.pivot_table(employees,index='department',aggfunc={'salary': 'sum', 'age': 'mean'})
```
