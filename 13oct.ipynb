```python
import numpy as np
import pandas as pd
```

```python
data=pd.Series([1,np.nan, 'hello', None])
print(data)
print()
print(data.isnull())
print()
print(data.dropna())
```

```python
df=pd.DataFrame([[1,np.nan,2,5],[2,3,5,np.nan],[8,np.nan,4,6],[1,6,np.nan,4],[np.nan,2,5,1]])
print(df)
print()
print(df.isnull())
print()
print(df.sum())
print()
print(df.isnull().sum())
```

```python
print(df.dropna())
print()
print(df.dropna(axis='columns'))

```

```python
df1=pd.DataFrame([[1,np.nan,2,5],[2,3,5,np.nan],[8,np.nan,4,6],[np.nan,np.nan,np.nan,np.nan],[np.nan,2,5,1]])
print(df1)
print()
print(df1.dropna(how='all'))
```

```python
print(df1.dropna(thresh=3))
```

```python
df2=pd.Series([1, np.nan, 2, None, 3, np.nan])
print(df2)
print()
print(df2.fillna(0))
print()
print(df2.ffill())
print()
print(df2.bfill())
```

```python
df3=pd.DataFrame({'Day': ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'],'Temperature': [22, 24, np.nan, 26, 25, np.nan, 23]})
print(df3)
print()
print(df3.fillna(df3['Temperature'].mean()))
print()
print(df3.interpolate())
```
