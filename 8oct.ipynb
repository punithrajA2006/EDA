```python
import numpy as np
import pandas as pd
```

```python

index = [['A',2010],['A',2020],['A',2025],['B',2010],['B',2020],['B',2025]]
population = [233,443,135,675,754,877]
i=pd.MultiIndex.from_tuples(index,names=['city','year'])
print(i)
pop = pd.Series(population,index=i)
print(pop)
```

```python
popu=pop.unstack()
print(popu)
pops=popu.stack()
print(pops)
```

```python
index1 = [['A',2010],['A',2020],['A',2025],['B',2010],['B',2020],['B',2025]]
population1 = [233,443,135,675,754,877]
i1=pd.MultiIndex.from_arrays([['A','A','A','B','B','B'],[2010,2020,2025,2010,2020,2025]],names=['city','year'])
print(i1)
pop1 = pd.Series(population1,index=i1)
print(pop1)
```

```python
i2=pd.MultiIndex.from_product([['A','B'],[2010,2020,2025]],names=['city','year'])
pop2 = pd.Series(population1,index=i2)
print(pop2)
```

```python
i3 =pd.MultiIndex(levels=[['A','B'],[2010,2020,2025]],codes=[[0,0,0,1,1,1],[0,1,2,0,1,2]],names=['city','year'])
pop3 = pd.Series(population1,index=i3)
print(pop3)
```
