```python
name=['alice','bob','charlie']
age=[25,30,35]
height=[5.5,6.0,5.8]

dat=np.zeros(3,dtype={'names':('name','age','height'),'formats':('U10','i4','f4')})
dat['name']=name
dat['age']=age
dat['height']=height
print(dat)

df5=pd.DataFrame(dat)
print(df5)
```

```python
ind=pd.Index([2,3,4,5,6,7])
print(ind)
print(ind.size,ind.shape,ind.ndim,ind.dtype)

print(ind[1])
print(ind[::2])
print(ind[1:4:2])

indA=pd.Index([1,3,5,7,9])
indB=pd.Index([2,3,5,7,11])
print(indA & indB)
print(indA | indB)
print(indA ^ indB)
print(indA.intersection(indB))

```

```python
area = pd.Series([603, 1484, 741, 428, 1855], index=['Mumbai', 'Delhi', 'Bengaluru', 'Chennai', 'Kolkata'])
population = pd.Series([12442338, 11034590, 8425970, 7036530, 4486676], index=['Mumbai', 'Delhi', 'Bengaluru', 'Chennai', 'Kolkata'])
states = pd.DataFrame({'area': area, 'population': population})
states['density'] = states['population'] / states['area']
print(states)
```

```python
indC = pd.Index([2,3,5,7,11])
indD = pd.Index([1,3,5,7,9])
print("Intersection:", indC.intersection(indD))
print("Union:", indC.union(indD))
print("Symmetric Difference:", indC.symmetric_difference(indD))
```
