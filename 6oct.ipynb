```python
import pandas as pd
import numpy as np
```

```python
data1 = pd.Series([1,np.nan,3,None,5],index=['a','b','c','d','e'])
data1.dropna()
print(data1)
```

```python
data2 = pd.Series([1,np.nan,3,None,5],index=['a','b','c','d','e'])
print(data2)
print(data2.isnull())
print(data2.notnull)
```

```python
print(data1)
print(data2)
```

```python
data3 = pd.DataFrame([[1,np.nan,2],[2,3,5],[np.nan,4,6]])
print(data3)
print(data3.dropna())
print(data3.dropna(axis='columns'))
print()
print(data3.dropna(axis='columns',how='all'))
print(data3.dropna(axis='rows',how='all'))
print(data3.dropna(axis='rows',thresh=3))
```

```python
data4= pd.DataFrame([[1,np.nan,2],[2,3,5],[np.nan,4,6]])
print(data4)
print()
print(data4.fillna(0))
print()
print(data4.ffill())
print()
print(data4.bfill())
```
