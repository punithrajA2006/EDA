```python
import numpy as np

student_names = np.array(["Aisha", "Ben", "Carlos", "Divya", "Ethan"])
scores = np.array([78, 92, 65, 88, 71])

rank = np.argsort(scores)
print(student_names[rank], scores[rank])
```

```python

X = np.random.rand(10, 2)
dist_sq = np.sum((X[:, np.newaxis, :] - X[np.newaxis, :, :]) ** 2, axis=-1)
nearest_neighbors = np.argsort(dist_sq, axis=1)[:, 1:3]
print(nearest_neighbors)
```
