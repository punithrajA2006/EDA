```python
import numpy as np

name=['Alice', 'Bob', 'Charlie', 'David', 'Eve']
age=[25, 30, 35, 40, 45]
marks=[84.5, 90.0, 95.5, 80.6, 75.0]

data=np.zeros(5, dtype=[('name', 'U10'), ('age', 'i4'), ('marks', 'f8')])
data['name']=name
data['age']=age
data['marks']=marks
print(data[data['age'] < 30]['name'])
```

```python
title = np.array(['The Great Gatsby', 'To Kill a Mockingbird','Pride and Prejudice', 'The Catcher in the Rye'])

author = np.array(['F. Scott Fitzgerald', 'Harper Lee', 'Jane Austen', 'J.D. Salinger'])    
price = np.array([104.49,407.95, 512.64, 129.99])
books=np.zeros(4, dtype=[('title', 'U30'), ('author', 'U20'), ('price', 'f8')])
books['title']=title
books['author']=author
books['price']=price

print(books)
```

```python
print(books['title'])
print(books[1])
print(books[books['price'] < 200]['title'])
```

```python
dtypea = {'names': ['title', 'author', 'price'], 'formats': ['U30', 'U20', 'f8']}
dtypeb = [('title', 'U30'), ('author', 'U20'), ('price', 'f8')]
print(np.dtype(dtypea) == np.dtype(dtypeb))
```
