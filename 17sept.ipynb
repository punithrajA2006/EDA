```python
import numpy as np

d= np.arange(25).reshape(5,5)
print(d)

e= d[:3,:4]
e[0,0] = 99
print(d)
print(e)
f= d[:3,:4].copy()
f[0,0] = 93
print(d)
print(f)
```

```python
np.random.seed(0)
a = np.random.randint(0,10, size =(2,3))
b = np.random.randint(0,10, size =(2,3))
print(a)
print(b)

w=np.vstack((a,b))
m=np.concatenate((a,b), axis=0)

print(w)
print(m)

o= np.hstack((a,b))
n= np.concatenate((a,b), axis=1)
print(o)
print(n)
```
