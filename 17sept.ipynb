```python
import numpy as np

arr = np.linspace(1,20,20,dtype=int)

print(arr)
print(arr.shape)
print(arr.ndim)
print(arr.size)
print(arr.dtype)
```

```python
import numpy as np
np.random.seed(1)
a = np.random.randint(0, 50, (4, 5))
print(a)
print(a.itemsize)
print(a.shape)
print(a.dtype)
print(a.ndim)
```

```python
np.random.seed(0)
b = np.random.randint(10, size=(3,4))
print(b[:2,:2])
print(b[:,::2])
print(b[:,::-1])
```
